Go to **Runtime** -> **Change runtime type** -> Select **T4 GPU** (100% Free) -> Click **Save**

```python
# ==============================================================================
# 🚀 BULLETPROOF COMFYUI + SDXL GOOGLE COLAB LAUNCHER
# ==============================================================================

import os, subprocess, threading, time, urllib.request

os.chdir('/content')

# 1. Clone ComfyUI
if not os.path.exists('ComfyUI'):
    print('⚡ [1/4] Cloning ComfyUI...')
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/comfyanonymous/ComfyUI.git'], check=True)

os.chdir('/content/ComfyUI')

# 2. Install Dependencies, Torch & Tunnel Tools
print('📦 [2/4] Installing dependencies & tunnel tools...')
subprocess.run(['pip', 'install', '-q', 'torch', 'torchvision', 'torchaudio', '--extra-index-url', 'https://download.pytorch.org/whl/cu121'])
subprocess.run(['pip', 'install', '-q', '-r', 'requirements.txt'])
subprocess.run(['pip', 'install', '-q', 'accelerate', 'transformers', 'safetensors', 'aiohttp', 'pycloudflared', 'pyngrok'])
subprocess.run(['npm', 'install', '-g', 'localtunnel'], check=False)

# 3. Download Photorealistic RealVisXL v5.0 Checkpoint (6.5 GB)
checkpoints_dir = 'models/checkpoints'
os.makedirs(checkpoints_dir, exist_ok=True)
model_url = 'https://huggingface.co/SG161222/RealVisXL_V5.0/resolve/main/RealVisXL_V5.0.safetensors'
model_path = os.path.join(checkpoints_dir, 'RealVisXL_V5.0.safetensors')

if not os.path.exists(model_path):
    print('📥 [3/4] Downloading RealVisXL v5.0 Checkpoint (6.5 GB)...')
    subprocess.run(['wget', '-c', '-q', '--show-progress', model_url, '-O', model_path])

# 4. Start ComfyUI Server on 0.0.0.0 with open CORS
print('🌐 [4/4] Starting ComfyUI server...')

def run_comfy():
    subprocess.run(['python', 'main.py', '--listen', '0.0.0.0', '--port', '8188', '--enable-cors-header', '*', '--highvram'])

threading.Thread(target=run_comfy, daemon=True).start()
time.sleep(6)

# Get Colab External IP for Localtunnel Password
try:
    colab_ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip()
except Exception:
    colab_ip = 'Check Colab IP'

print('\n' + '=' * 65)
print('🎉 COMFYUI IS RUNNING ON GOOGLE COLAB GPU!')
print('=' * 65)
print(f'🔑 LOCALTUNNEL PASSWORD (if prompted on page): {colab_ip}')
print('-' * 65)

# Start Localtunnel (Backup 1)
subprocess.Popen(['npx', 'localtunnel', '--port', '8188'])

# Start Cloudflare Tunnel (Backup 2)
try:
    from pycloudflared import try_cloudflare
    cf_url = try_cloudflare(port=8188)
    print('👉 Cloudflare Link:', cf_url)
except Exception as e:
    pass

print('=' * 65 + '\n')
```
